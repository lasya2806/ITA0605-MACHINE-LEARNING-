```python
# EXPERIMENT 20
# Future Sales Prediction using Linear Regression

import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

# Advertising Budget, Number of Salespersons
X = np.array([
    [20, 2],
    [25, 2],
    [30, 3],
    [35, 3],
    [40, 4],
    [45, 4],
    [50, 5],
    [55, 5],
    [60, 6],
    [65, 6]
])

# Monthly Sales
y = np.array([
    250,
    290,
    340,
    390,
    450,
    510,
    580,
    640,
    710,
    780
])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

model = LinearRegression()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Actual Sales:", y_test)

print("Predicted Sales:", y_pred)

print("Mean Squared Error:",
      mean_squared_error(y_test, y_pred))

print("R2 Score:",
      r2_score(y_test, y_pred))

new_data = np.array([[48, 5]])

prediction = model.predict(new_data)

print()
print("Advertising Budget: 48,000")
print("Number of Salespersons: 5")

print("Predicted Future Sales:",
      round(prediction[0], 2),
      "Units")

plt.scatter(y_test, y_pred)

plt.xlabel("Actual Sales")
plt.ylabel("Predicted Sales")

plt.title("Future Sales Prediction")

plt.show()

plt.scatter(X[:, 0], y)

plt.xlabel("Advertising Budget")
plt.ylabel("Sales")

plt.title("Advertising Budget vs Sales")

plt.show()
```
