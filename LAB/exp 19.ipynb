```python
# EXPERIMENT 19
# Bank Loan Prediction using Naive Bayes

import numpy as np
import matplotlib.pyplot as plt

from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split

from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import ConfusionMatrixDisplay

# Annual Income, Down Payment, Credit Score
X = np.array([
    [14, 3.0, 830],
    [12, 2.5, 810],
    [10, 2.0, 780],
    [8, 1.8, 740],
    [7, 1.5, 690],
    [6, 1.2, 660],
    [5, 1.0, 630],
    [11, 2.2, 790],
    [4, 0.8, 600],
    [15, 3.5, 850]
])

# 1 = Approved
# 0 = Rejected

y = np.array([
    1,
    1,
    1,
    1,
    0,
    0,
    0,
    1,
    0,
    1
])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

model = GaussianNB()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Actual Values:", y_test)

print("Predicted Values:", y_pred)

print("Accuracy:",
      accuracy_score(y_test, y_pred))

new_applicant = np.array([
    [11, 2.4, 800]
])

prediction = model.predict(new_applicant)

if prediction[0] == 1:
    print("Predicted Loan Status: Approved")
else:
    print("Predicted Loan Status: Rejected")

cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Rejected", "Approved"]
)

disp.plot()

plt.title("Bank Loan Prediction using Naive Bayes")

plt.show()
```
