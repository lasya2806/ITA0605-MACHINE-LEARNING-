```python
# EXPERIMENT 10
# Expectation Maximization Algorithm

import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

# Dataset
X = np.array([
    [2, 65],
    [3, 70],
    [4, 75],
    [5, 80],
    [6, 85],
    [7, 88],
    [8, 90],
    [9, 93],
    [10, 95],
    [11, 98]
])

# EM using Gaussian Mixture Model
model = GaussianMixture(n_components=2, random_state=42)

# Train the model
model.fit(X)

# Cluster prediction
clusters = model.predict(X)

print("Cluster Labels:")
print(clusters)

# New sample
new_sample = np.array([[7, 89]])

# Predict cluster for new sample
new_cluster = model.predict(new_sample)

print("\nNew Sample:", new_sample[0])
print("Predicted Cluster:", new_cluster[0])

# Graph
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=clusters,
    s=100
)

# Plot new sample
plt.scatter(
    new_sample[:, 0],
    new_sample[:, 1],
    marker="X",
    s=250
)

plt.xlabel("Study Hours")
plt.ylabel("Attendance (%)")
plt.title("Expectation Maximization Clustering")
plt.show()
```
